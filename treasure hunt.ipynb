```python
map_rooms={
    "A":["B","C"],
    "B":["D"],
    "C":["D"],
    "D":["G"],
    "G":[]
}
```

```python
def tressure_hunt(start,goal):
    to_do=[start]
    visited=[]
    order=[]
    while to_do:
        room = to_do.pop(0)
        if room in visited:
            continue
        visited.append(room)
        order.append(room)
        if room==goal:
            return order
        for nxt in map_rooms[room]:
            to_do.append(nxt)
    return order
order=tressure_hunt("A","G")
print("BFS ORDER:",order)
print("length of path:",len(order))
        
```
